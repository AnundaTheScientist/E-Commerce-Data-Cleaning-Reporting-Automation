## 1. Business Scenario

> We are simulating a small-to-medium online retailer that receives recurring transaction exports and currently relies on manual preparation before producing management reports.


## 2. Core Problem

Recurring e-commerce transaction data requires **manual cleaning, validation, and preparation** before it can reliably support business reporting.


## 3. Business Impact

Left unaddressed, the manual process can create:

- Unnecessary reporting effort
- Inconsistent calculations
- Duplicated or missing records
- Unreliable KPIs
- Difficulty reproducing previous reports
- Dependence on individual spreadsheet work


## 4. Business Objective

### **Build a repeatable data-quality and reporting pipeline that transforms messy e-commerce transaction exports into validated business information and management-ready reports with minimal manual intervention.**


## 5. Business Questions

The final report is designed to answer the following questions, grouped by business area.


### 5.1 Sales

1. What is total gross revenue?
2. What is net revenue after returns?
3. How many orders were processed?
4. How many units were sold?
5. How does performance change over time?


### 5.2 Products

1. Which products generate the most revenue?
2. Which products sell the most units?
3. Which products have significant returns?


### 5.3 Customers

1. How many customers purchased?
2. What is the average order value?
3. Which customers contribute significant revenue?


### 5.4 Geography

1. Which countries/markets generate the most revenue?
2. How does sales performance vary geographically?


### 5.5 Data Quality

1. How many records were received?
2. How many issues were detected?
3. How many duplicates were found?
4. How many cancellations/returns were identified?
5. How many records were excluded?
6. Why were records excluded?


### 5.6 Automation

1. Can the same process be run against future exports without manually repeating every cleaning step?
