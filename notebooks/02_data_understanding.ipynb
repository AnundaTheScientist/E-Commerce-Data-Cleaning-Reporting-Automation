## 2.2.6 Missing-Value Profile

A missing-value profile is used to quantify the completeness of each
field in the raw transaction data.

This analysis is descriptive only. No records will be removed or
imputed at this stage.

### Missing-Value Findings

The raw data contains missing values in [fields].

The most significant observed missingness occurs in [field], with
approximately [X]% of records missing in [period] and [Y]% in
[period].

These findings indicate that customer-level and/or product-level
analysis may require explicit handling rules during Data Preparation.

No records have been removed or imputed during this stage.

```python
missing_profiles = []

for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    profile = pd.DataFrame({
        "Column": data.columns,
        "Total Records": len(data),
        "Missing Records": data.isna().sum().values
    })

    profile["Missing %"] = (
        profile["Missing Records"] / profile["Total Records"] * 100
    ).round(2)

    profile["Present Records"] = (
        profile["Total Records"] - profile["Missing Records"]
    )

    profile["Present %"] = (
        profile["Present Records"] / profile["Total Records"] * 100
    ).round(2)

    profile.insert(0, "Reporting Period", sheet)

    missing_profiles.append(profile)

missing_profile = pd.concat(
    missing_profiles,
    ignore_index=True
)

missing_profile

missing_profile.sort_values(
    ["Reporting Period", "Missing %"],
    ascending=[True, False]
)
```

### Missing-Value Findings

The initial missing-value profile identified two fields with missing
records across both reporting periods:

- `Customer ID` has the highest level of missingness, with 20.54% of
  records missing in 2009-2010 and 24.93% missing in 2010-2011.
- `Description` has comparatively low missingness, with 0.56% missing
  in 2009-2010 and 0.27% missing in 2010-2011.

The fields `Invoice`, `StockCode`, `Quantity`, `InvoiceDate`, `Price`,
and `Country` contain no missing values in either reporting period.

At this stage, no records are removed and no missing values are
imputed. The missing fields will be investigated further to determine
their business impact and appropriate treatment during the Data
Preparation phase.

### Initial Data-Quality Implications

The missing `Customer ID` values may limit customer-level analysis,
but the associated transaction records may still contain useful
information for product, geographic, transaction, and revenue analysis.

Missing `Description` values require further investigation before a
cleaning rule is established.

Missing-value analysis alone does not establish overall data quality.
Additional profiling is required for duplicate records, cancellations
and returns, negative or zero quantities, unusual prices, transaction
identifiers, and categorical inconsistencies.

## 2.2.7 Duplicate & Transaction-Level Investigation

Repeated invoice numbers are not automatically treated as duplicates
because a single invoice may contain multiple product lines.

The objective of this investigation is to distinguish normal
transaction structure from potentially duplicated records.

```python
duplicate_profiles = []

for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    exact_duplicates = data.duplicated().sum()
    unique_invoices = data["Invoice"].nunique()
    total_records = len(data)

    duplicate_profiles.append({
        "Reporting Period": sheet,
        "Total Records": total_records,
        "Unique Invoices": unique_invoices,
        "Repeated Invoice Records": total_records - unique_invoices,
        "Exact Duplicate Rows": exact_duplicates
    })

duplicate_profile = pd.DataFrame(duplicate_profiles)

duplicate_profile
```

```python
invoice_line_profile = []

for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    invoice_counts = data.groupby("Invoice").size()

    invoice_line_profile.append({
        "Reporting Period": sheet,
        "Invoices": len(invoice_counts),
        "Invoices With Multiple Lines": (invoice_counts > 1).sum(),
        "Maximum Lines Per Invoice": invoice_counts.max(),
        "Average Lines Per Invoice": round(invoice_counts.mean(), 2)
    })

invoice_line_profile = pd.DataFrame(invoice_line_profile)

invoice_line_profile
```

### Duplicate and Transaction-Level Findings

The raw workbook contains repeated invoice numbers across multiple
transaction lines. This is expected because the dataset is structured
at the transaction-line level rather than one row per invoice.

In the 2009-2010 reporting period, 28,816 unique invoices are
represented across 525,461 transaction-line records. In the 2010-2011
reporting period, 25,900 unique invoices are represented across
541,910 transaction-line records.

Most invoices contain multiple transaction lines. The average number of
lines per invoice is 18.24 in 2009-2010 and 20.92 in 2010-2011.

The profile also identified 6,865 exact duplicate rows in 2009-2010 and
5,268 exact duplicate rows in 2010-2011.

Repeated invoice numbers are therefore not treated as duplicates.
Exact duplicate rows require further investigation before a cleaning
rule is established.

### Data-Quality Implications

The transaction grain must be preserved when designing the cleaning
pipeline. Invoice-level aggregation should be performed explicitly
rather than assuming one row represents one order.

Exact duplicate rows are a potential source of double-counting and may
affect revenue, quantity, and other transaction-level metrics. Their
treatment will be determined after further investigation.

## 2.2.8 Returns, Cancellations and Negative Values

E-commerce transaction exports may contain cancelled orders, returns,
adjustments or other non-standard transactions.

This investigation identifies negative and zero quantities and examines
their relationship with invoice identifiers before any treatment rule is
defined.

```python
transaction_quality_profiles = []

for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    negative_quantity = (data["Quantity"] < 0).sum()
    zero_quantity = (data["Quantity"] == 0).sum()
    negative_price = (data["Price"] < 0).sum()
    zero_price = (data["Price"] == 0).sum()

    cancellation_invoices = (
        data["Invoice"]
        .astype(str)
        .str.upper()
        .str.startswith("C")
        .sum()
    )

    transaction_quality_profiles.append({
        "Reporting Period": sheet,
        "Negative Quantity Rows": negative_quantity,
        "Zero Quantity Rows": zero_quantity,
        "Negative Price Rows": negative_price,
        "Zero Price Rows": zero_price,
        "Cancellation Invoice Rows": cancellation_invoices
    })

transaction_quality_profile = pd.DataFrame(
    transaction_quality_profiles
)

transaction_quality_profile
```

```python
for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    cancellation_mask = (
        data["Invoice"]
        .astype(str)
        .str.upper()
        .str.startswith("C")
    )

    negative_quantity_mask = data["Quantity"] < 0

    print(f"\n{sheet}")
    print("-" * len(sheet))

    print(
        "Negative-quantity rows:",
        negative_quantity_mask.sum()
    )

    print(
        "Negative-quantity rows with cancellation invoice:",
        (
            negative_quantity_mask &
            cancellation_mask
        ).sum()
    )

    print(
        "Negative-quantity rows without cancellation invoice:",
        (
            negative_quantity_mask &
            ~cancellation_mask
        ).sum()
    )
```

### Returns, Cancellations and Negative-Value Findings

The transaction profile identified 12,326 negative-quantity rows in
2009-2010 and 10,624 in 2010-2011. No zero-quantity rows were found in
either reporting period.

Most negative-quantity rows are associated with cancellation-style
invoice identifiers. In 2009-2010, 10,205 of 12,326 negative-quantity
rows have cancellation-style invoice identifiers. In 2010-2011,
9,288 of 10,624 negative-quantity rows have cancellation-style invoice
identifiers.

However, 2,121 negative-quantity rows in 2009-2010 and 1,336 in
2010-2011 do not have cancellation-style invoice identifiers. These
records require further investigation before a treatment rule is
defined.

The dataset also contains 3,687 zero-price rows in 2009-2010 and 2,515
in 2010-2011. Five rows contain negative prices across the two periods.

No records are removed at this stage. The findings indicate that the
final reporting model should distinguish positive sales activity from
returns or cancellations rather than treating all negative quantities
as invalid data.

Zero-price and negative-price transactions will be investigated further
before cleaning rules are established.

## 2.2.11 Non-Standard StockCode Investigation

The StockCode field contains both numeric-leading and alphabetic-leading
values.

Because some alphabetic StockCodes may represent operational,
adjustment, or other non-standard transactions, their associated
descriptions, quantities, prices and invoice patterns will be examined
before transaction-classification rules are established.

```python
for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    data["StockCode_Str"] = data["StockCode"].astype(str).str.strip()

    non_numeric = data[
        ~data["StockCode_Str"].str[0].str.isdigit()
    ].copy()

    profile = (
        non_numeric
        .groupby("StockCode_Str")
        .agg(
            Rows=("StockCode_Str", "size"),
            Descriptions=("Description", "nunique"),
            Negative_Quantity=("Quantity", lambda x: (x < 0).sum()),
            Zero_Price=("Price", lambda x: (x == 0).sum()),
            Positive_Price=("Price", lambda x: (x > 0).sum()),
            Missing_Customer_ID=("Customer ID", lambda x: x.isna().sum())
        )
        .sort_values("Rows", ascending=False)
    )

    print(f"\n{sheet}")
    print("=" * 70)

    display(profile)
```

## 2.2.11 — Final investigation

We have identified the codes, but we haven't yet established their exact descriptions and business behavior across the whole dataset.

```python
for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    data["StockCode_Str"] = data["StockCode"].astype(str).str.strip()

    non_numeric = data[
        ~data["StockCode_Str"].str[0].str.isdigit()
    ].copy()

    summary = (
        non_numeric
        .groupby("StockCode_Str")
        .agg(
            Rows=("StockCode_Str", "size"),
            Description=("Description", lambda x: " | ".join(
                x.dropna().astype(str).unique()[:3]
            )),
            Min_Quantity=("Quantity", "min"),
            Max_Quantity=("Quantity", "max"),
            Min_Price=("Price", "min"),
            Max_Price=("Price", "max"),
            Missing_Customer_ID=("Customer ID", lambda x: x.isna().sum()),
            Invoice_Prefixes=("Invoice", lambda x: ", ".join(
                sorted(
                    x.astype(str)
                    .str[0]
                    .dropna()
                    .unique()
                )
            ))
        )
        .sort_values("Rows", ascending=False)
    )

    print(f"\n{sheet}")
    print("=" * 100)

    display(summary)
```

```python
for sheet in excel_file.sheet_names:
    data = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )

    data["LineValue"] = data["Quantity"] * data["Price"]

    print(f"\n{sheet}")
    print("=" * 70)

    print("Total raw line value:", data["LineValue"].sum())
    print("Minimum line value:", data["LineValue"].min())
    print("Maximum line value:", data["LineValue"].max())

    print("\nLine-value summary:")
    display(
        data["LineValue"].describe()
    )

    print("\nNegative line values:", (data["LineValue"] < 0).sum())
    print("Zero line values:", (data["LineValue"] == 0).sum())
    print("Positive line values:", (data["LineValue"] > 0).sum())

    print("\nTop 15 transactions by absolute line value:")
    display(
        data.loc[
            data["LineValue"].abs().nlargest(15).index,
            [
                "Invoice",
                "StockCode",
                "Description",
                "Quantity",
                "Price",
                "LineValue",
                "Customer ID",
                "Country"
            ]
        ].sort_values(
            "LineValue",
            key=lambda x: x.abs(),
            ascending=False
        )
    )
```
